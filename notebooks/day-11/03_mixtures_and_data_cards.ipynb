### Reference answer and mechanism

```python
inverse=[1/length for length in lengths]
balanced=data.mixture_exposure(inverse,lengths)
assert np.allclose(balanced,[.5,.5])
print("Inverse-length example proportions:",np.array(inverse)/sum(inverse))
print("Token shares:",balanced)
```

### Reference explanation

### Reference answer and mechanism

Sample short examples more often: weights proportional to inverse length. The resulting example distribution changes, so it must be declared and evaluated.
