````python
for change in ("cycle", "missing-parent", "changed-panel"):
    broken = deepcopy(cards)
    if change == "cycle":
        broken[0]["parent"] = "toy-rl"
    elif change == "missing-parent":
        broken[1]["parent"] = "absent-checkpoint"
    else:
        broken[2]["evaluation_sha256"] = "different-panel"
    try:
        validate_genealogy(broken)
    except ValueError as error:
        print(change, "→", error)
print("For real cards, compute hashes from actual weights/data and link measured reports.")
````

### Reference explanation

Each intentional cycle, missing parent and changed evaluation identity raises a validation error. The fixture validates graph and panel structure; a label-derived hash is not a checksum of a trained checkpoint.
