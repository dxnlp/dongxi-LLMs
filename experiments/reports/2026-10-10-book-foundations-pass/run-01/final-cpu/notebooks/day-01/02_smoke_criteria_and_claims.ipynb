```python
from dongxi_llms.course_foundations import assess_smoke
cases = {'passes': (0, [2.1, 1.9], 31, 25),
         'unsafe reserve': (0, [2.1, 1.9], 20, 25),
         'no measured loss': (0, [], 31, 25),
         'nonfinite loss': (0, [float('nan')], 31, 25),
         'failed exit': (1, [2.1, 1.9], 31, 25)}
results = {name: assess_smoke(*args) for name, args in cases.items()}
assert results['passes']['accepted']
assert all(not result['accepted'] for name, result in results.items() if name != 'passes')
for name, result in results.items(): print(name, result['criteria'], result['accepted'])
```

```python
names = list(results)
matrix = torch.tensor([[int(v) for v in results[n]['criteria'].values()] for n in names])
fig, ax = plt.subplots(figsize=(8, 3.5))
ax.imshow(matrix, cmap='Greys', vmin=0, vmax=1, aspect='auto')
ax.set_xticks(range(3), ['Exit zero', 'Finite losses', 'Sampled reserve'])
ax.set_yticks(range(len(names)), names)
for i in range(len(names)):
    for j in range(3): ax.text(j, i, 'pass' if matrix[i,j] else 'fail / absent', ha='center', va='center', color='white' if matrix[i,j] else 'black')
ax.set_title('Synthetic criterion controls; black = passed')
plt.show()
```
