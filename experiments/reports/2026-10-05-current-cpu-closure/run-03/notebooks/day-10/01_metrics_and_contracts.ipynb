## Predict 2

With ten independent candidates and two verified successes, predict how allowing one, three, or ten attempts changes the event being scored.

Pause and explain your reasoning before running the next cell.

### Reference answer and mechanism

pass@k is the probability that a chosen k-subset contains at least one success. It changes the sampling budget; it does not tell a deployed model which answer to select.

```python
ks = np.arange(1,11)
estimated = [ev.pass_at_k(10,2,int(k)) for k in ks]
assert math.isclose(estimated[2], 1-math.comb(8,3)/math.comb(10,3))
fig, ax = plt.subplots(figsize=(7,3.5))
ax.plot(ks,estimated,"o-",color="#346c9c")
ax.set(xlabel="Allowed candidates k",ylabel="Estimated pass@k",title="Same observed pool, different attempt budget",ylim=(0,1.05))
plt.show()
```

## Predict 3

Now imagine the ten candidates are ten copies of one output. Why is the same estimator interpretation inappropriate?

Pause and explain your reasoning before running the next cell.

### Reference answer and mechanism

The combinatorial computation still counts subsets of the recorded pool, but repeated copies do not model new independent draws. An apparent budget increase does not supply independent exploration.

```python
n,c,k = 10,2,3
print("Finite-pool accounting:", ev.pass_at_k(n,c,k))
try: ev.pass_at_k(2,1,3)
except ValueError as exc: print("Rejected unsupported k:", exc)
assert ev.pass_at_k(10,0,3) == 0 and ev.pass_at_k(10,10,3) == 1
```
